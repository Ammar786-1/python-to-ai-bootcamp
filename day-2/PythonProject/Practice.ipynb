```python
import pandas as pd 
import numpy as pb
print("Libraries imported")
```

```python
Score= pb.array([56, 45, 40, 28, 100])
print("Score:", Score)
print("Sum:", pb.sum(Score))
print("Mean:", pb.mean(Score))
print("Standard Deviation:", pb.std(Score))
print("Median:", pb.median(Score))
print("Max:", pb.max(Score))
print("Min:", pb.min(Score))
```

```python
print("Survival count:") 
display(df["Survived"].value_counts())
print("Survival percentage%:")
display(
    df["Survived"].value_counts(normalize=True) * 100
)
```

```python
print("Passengers count:")
display(df["Sex"].value_counts())
 
print("Genders percentage:")
 
display(
    df["Sex"].value_counts(normalize=True) * 100
)
```

```python
countPersons = (df["Age"] > 25).sum()

print(countPersons)
```

```python
print("Males older than 25:")
print(df[(df["Age"] > 25) & (df["Sex"] == "male")]["Name"])

print("\nFemales older than 25:")
print(df[(df["Age"] > 25) & (df["Sex"] == "female")]["Name"])
```
